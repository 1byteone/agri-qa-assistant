## 01 基础概念

### 1.1 什么是 Prompt Engineering

**Prompt Engineering** 是指在**不修改模型权重**的前提下，通过设计、优化和迭代输入文本，引导 LLM 生成期望输出的工程学科。

**核心比喻**：把 LLM 当作实习生，Prompt Engineer 是项目经理——不修改大脑结构，而是通过清晰的任务描述、背景资料、成功范例和质量标准，确保高效完成任务。

### 1.2 核心术语

| 术语 | 说明 |
|------|------|
| Prompt | 输入给模型的文本指令 |
| System Prompt | 设定身份、规则、风格的顶层指令 |
| Context Window | 模型单次能处理的最大 Token 数 |
| Token | 基本处理单位，≈ 0.75 英文单词 |
| Hallucination | 模型生成看似合理但事实错误的内容 |
| Few-shot | 在 Prompt 中提供若干输入-输出示例 |
| Zero-shot | 不提供示例，直接要求完成任务 |
| RAG | 从外部知识库检索内容注入 Prompt |
| CoT | 要求模型逐步推理再给出答案 |
| Temperature | 控制输出随机性的参数

### 1.3 发展脉络

```
2020  GPT-3 发布，In-Context Learning 概念确立
2021  Chain-of-Thought (CoT) 提出
2022  ChatGPT 发布，RLHF 让模型更安全
2023  GPT-4 / Claude 3 / Llama 2 多模态与长上下文
2024  o1 / Claude 3.5 Sonnet 推理模型兴起
2025  Prompt Caching / MCP / Agentic Workflow 成为生产标配
```

### 1.4 Prompt Engineering 与模型优化的关系

| 技术 | 成本 | 速度 | 深度 | 适用场景 |
|------|------|------|------|----------|
| Prompt Engineering | 极低 | 分钟级 | 浅层控制 | 快速迭代 |
| PEFT / LoRA | 中 | 小时级 | 中等 | 稳定领域风格注入 |
| Full Fine-tuning | 高 | 天级 | 深层 | 全量领域知识迁移 |
| RLHF | 极高 | 周/月级 | 基础对齐 | 模型厂商出厂配置 |

**结论**：Prompt Engineering 是离应用最近、最灵活的层，应作为默认起点。

### 1.5 主流模型特性对比

| 维度 | GPT-4o / GPT-5 | Claude 3.5 / 4 | Llama 3 / Qwen | DeepSeek |
|------|-----------------|-----------------|----------------|----------|
| 上下文长度 | 128K / 1M | 200K | 8K-128K | 64K-128K |
| 指令遵循 | 极强 | 极强 | 中等 | 强 |
| 推理能力 | o1/o3 极强 | Sonnet 强 | 依赖规模 | 数学/代码强 |
| 中文能力 | 强 | 强 | Qwen 极强 | 极强 |
| 本地部署 | 不支持 | 不支持 | 支持 | 支持 |
| 成本 | 中高 | 中 | 低 | 低 |

## 02 设计框架

### 2.1 ICIO 框架（最简单）

| 要素 | 说明 |
|------|------|
| Instruction | 明确的任务动词 |
| Context | 任务相关的前置信息 |
| Input Data | 待处理的具体内容 |
| Output Indicator | 期望的输出格式 |

### 2.2 CRISPE 框架（强调角色）

- **C**apacity and Role：模型扮演什么角色
- **R**elevant Background：提供充足背景
- **I**nstruction：具体任务描述
- **S**tyle and Format：输出风格与格式
- **P**rovide Examples：给出示例
- **E**valuate and Refine：评估并迭代优化

### 2.3 BROKE 框架（迭代优化）

- **B**ackground：背景信息
- **R**ole：角色定义
- **O**bjectives：具体目标
- **K**ey Points：关键关注点
- **E**volve：根据输出改进 Prompt

### 2.4 RASCEF 框架（复杂任务）

- **R**ole：角色
- **A**ction：具体行动
- **S**teps：详细步骤
- **C**ontext：背景
- **E**xamples：示例
- **F**ormat：输出格式

### 2.5 七大要素（综合版）

1. **Role（角色）**：明确具体
2. **Context（背景）**：前置关键信息
3. **Task（任务）**：用指令动词开头
4. **Examples（示例）**：质量 > 数量
5. **Format（格式）**：JSON / Markdown / XML
6. **Quality & Constraints（质量与约束）**：量化指标 + 负向约束
7. **Thinking Time（思考时间）**：要求 `<thinking>` + `<answer>`

### 2.6 代码示例：Python + TypeScript 调用

以下代码展示了如何用 OpenAI 和 Anthropic SDK 发送结构化的 Prompt。

````python
# OpenAI Python SDK 示例
from openai import OpenAI

client = OpenAI()  # 需要设置 OPENAI_API_KEY 环境变量

response = client.responses.create(
    model="gpt-4o",
    input=[
        {
            "role": "developer",
            "content": "你是一个专业的代码审查员，擅长发现 Python 代码中的问题。"
        },
        {
            "role": "user",
            "content": "请审查以下代码：\n\n```python\ndef add(a, b):\n    return a + b\n```"
        }
    ]
)

print(response.output_text)
````

````python
# Anthropic Python SDK 示例
from anthropic import Anthropic

client = Anthropic()  # 需要设置 ANTHROPIC_API_KEY 环境变量

message = client.messages.create(
    model="claude-sonnet-4-20250514",
    max_tokens=1024,
    system="你是一个专业的代码审查员，擅长发现 Python 代码中的问题。",
    messages=[
        {
            "role": "user",
            "content": "请审查以下代码：\n\n```python\ndef add(a, b):\n    return a + b\n```"
        }
    ]
)

print(message.content[0].text)
````

## 03 进阶技术

### 3.1 Few-Shot Learning

在 Prompt 中提供若干输入-输出示例，让模型模仿模式。

```
Text: (lawrence bounces) all over the stage, dancing...
Sentiment: positive

Text: despite all evidence to the contrary, this clunker...
Sentiment: negative

Text: i'll bet the video game is a lot more fun than the film.
Sentiment:
```

### 3.2 Chain-of-Thought（链式思考）

要求模型展示逐步推理过程，显著提升数学、逻辑、代码任务准确率。

**Zero-shot CoT 模板**：
```
Question: {复杂问题}
Answer: Let's think step by step.
```

**Few-shot CoT 模板**：
```
Question: {问题}
Answer: 推理步骤 1...
推理步骤 2...
所以答案是：{答案}
===
Question: {新问题}
Answer:
```

### 3.3 Tree of Thoughts（思维树）

在每一步生成多个候选思路，形成树结构，用 BFS/DFS 搜索最优路径。

适用场景：复杂规划、数学证明、创意写作。

### 3.4 ReAct（推理 + 行动）

交替进行思考（Thought）和工具调用（Action），将推理过程显式化。

```
Thought: 我需要查一下当前北京的天气。
Action: WeatherAPI("Beijing")
Observation: 15°C，晴
Thought: 用户问的是穿衣建议，我需要基于温度给出推荐。
Answer: 建议穿薄外套...
```

### 3.5 RAG（检索增强生成）

```
1. 用户提问
2. 检索系统从知识库找出最相关的 Top-K 文档片段
3. 将片段作为 Context 注入 Prompt
4. 模型生成基于检索内容的答案
```

**格式建议**：
```
Evidence: {检索到的段落}
Question: {用户问题}
Answer:
```

### 3.6 Agent 工作流模式

````python
# RAG 完整示例（简化版）
# 需要安装: pip install openai chromadb

import chromadb
from openai import OpenAI

# 初始化客户端
client = OpenAI()
chroma_client = chromadb.Client()
collection = chroma_client.get_or_create_collection("knowledge_base")

# 模拟知识库文档
documents = [
    "公司年假政策：员工工作满 1 年享受 5 天年假，满 5 年享受 10 天，满 10 年享受 15 天。",
    "请假流程：在 OA 系统提交申请，直属上级审批，超过 3 天需 HR 备案。",
    "远程办公政策：每周二、周四可申请远程办公，需提前一天在系统提交。"
]

# 生成 embeddings 并存储（简化）
for i, doc in enumerate(documents):
    embedding = client.embeddings.create(
        model="text-embedding-3-small",
        input=doc
    ).data[0].embedding
    collection.add(
        documents=[doc],
        embeddings=[embedding],
        ids=[f"doc_{i}"]
    )

# 检索并问答
def rag_query(question: str):
    # 1. 检索
    query_embedding = client.embeddings.create(
        model="text-embedding-3-small",
        input=question
    ).data[0].embedding
    
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=2
    )
    
    context = "\n\n".join(results["documents"][0])
    
    # 2. 构建 Prompt
    prompt = f"""
    # Identity
    你是公司内部智能助手，专门回答员工关于公司政策的问题。

    # Instructions
    - 只基于提供的上下文回答
    - 如果上下文中没有答案，说"我暂时没有这方面的信息，建议咨询 HR"
    - 引用具体政策条款

    # Context
    {context}

    # Question
    {question}
    """
    
    # 3. 生成答案
    response = client.responses.create(
        model="gpt-4o",
        input=prompt
    )
    return response.output_text

# 测试（取消注释并填入 API Key 后运行）
# answer = rag_query("工作 3 年有多少天年假？")
# print(answer)
````

### 3.7 技术选型决策树

```
任务复杂吗？
├── 否 → Zero-shot + 明确指令
└── 是
    ├── 需要外部知识？ → RAG
    ├── 需要多步推理？ → CoT / ToT
    ├── 需要工具调用？ → ReAct / Agent
    ├── 步骤固定？ → Prompt Chaining
    └── 需多视角验证？ → Parallelization / Voting
```

## 04 模板实例

> 以下模板均来自开源项目 awesome-chatgpt-prompts 及社区贡献，可直接复用。

### 4.1 编程辅助类模板

**代码审查员**：
```
You are an expert code reviewer with deep knowledge of {language} best practices,
security vulnerabilities, and performance optimization.

Review the following code and provide:
1. Potential bugs or edge cases
2. Security vulnerabilities
3. Performance improvements
4. Code style and readability suggestions
5. A corrected version of the code
```

### 4.2 写作与内容类模板

**文章改写**：
```
You are an academic writing assistant with expertise in {field}.

Rewrite the following text to:
- Improve clarity and coherence
- Use formal academic tone
- Remove colloquialisms
- Ensure logical flow between paragraphs

Original text: {text}
Target length: {word_count} words
```

### 4.3 结构化 JSON 模板

```json
{
  "version": "1.0",
  "metadata": {
    "name": "code-reviewer",
    "model": "gpt-4o",
    "temperature": 0.2,
    "max_tokens": 2048
  },
  "system_prompt": {
    "identity": "You are an expert code reviewer.",
    "capabilities": ["Identify bugs", "Detect security issues"],
    "constraints": ["Do not modify code directly"]
  },
  "user_prompt_template": "Review: {code}",
  "output_format": {
    "type": "markdown",
    "sections": ["bugs", "security", "performance"]
  }
}
```

## 05 最佳实践

### 5.1 OpenAI 官方建议

- 使用 `developer` 消息传递系统指令（优先级最高）
- 将 Prompt 存入代码库，而非使用可复用的 Prompt 对象
- 使用类型化函数参数传动态值
- 在改生产 Prompt 前，先补测试和评估集

### 5.2 Anthropic 官方建议

- **简洁性**：优先用最简方案，仅在需要时增加复杂度
- **透明性**：显式展示 Agent 的规划步骤
- **ACI 质量**：工具文档和测试投入 ≥ Prompt 优化投入

### 5.3 多模型特定最佳实践

| 任务特征 | 首选模型 | 原因 |
|----------|----------|------|
| 精确指令遵循 | GPT-4o | 指令遵循能力强 |
| 长文档分析 | Claude 3.5 Sonnet | 200K 上下文 |
| 复杂推理/数学 | o1 / DeepSeek-R1 | 推理模型 |
| 代码生成 | GPT-4o | 代码能力强 |
| 中文写作 | DeepSeek / Qwen | 中文优化好 |
| 成本敏感 | Qwen / DeepSeek | API 价格低 |
| 数据隐私 | Llama 3 / Qwen | 本地部署 |

## 06 反模式与练习题

### 6.1 常见错误速查

| 反模式 | 修正 |
|--------|------|
| 模糊指令 | 用具体动词和量化指标 |
| 否定式指令 | 改为正面指令 |
| 超长输入 | 用 RAG 或摘要 |
| 无格式约束 | 明确 JSON / Markdown |
| 示例质量差 | 示例必须与目标输出一致 |
| Prompt 硬编码 | 存入代码库 + CI/CD |

### 6.2 练习题
